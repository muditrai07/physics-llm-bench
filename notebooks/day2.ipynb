```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Arrays and vectorized math: no loops
M = np.linspace(1, 100, 5)        # 5 masses from 1 to 100 GeV
print(M, M**2, np.sqrt(M))

# 2. Broadcasting: an array combined with a number works elementwise
m = 0.5
print(M**2 - 4*m**2)

# 3. Masks: pick elements by a condition
print(M[M > 30])

# 4. Comparing floats: never use == on computed numbers
print(np.isclose(0.1 + 0.2, 0.3), 0.1 + 0.2 == 0.3)
```
