```python
m_phi, m_f, y = sp.symbols("m_phi m_f y", positive=True)
beta = sp.sqrt(1 - 4*m_f**2/m_phi**2)
Gamma = y**2 * m_phi * beta**3 / (8*sp.pi)
```

```python
sp.series(Gamma, m_f, 0, 4)

sp.solve(sp.Eq(beta, 0), m_f)

sp.simplify(Gamma.subs(m_f, 0))

```

```python
f = sp.lambdify((m_phi, m_f, y), Gamma, "numpy")
Mscalar=100
massf=np.linspace(1,49,50)
width_scalar_to_fermions(Mscalar,massf,1)
np.allclose(f(Mscalar,massf,1),width_scalar_to_fermions(Mscalar,massf,1)
)
```
